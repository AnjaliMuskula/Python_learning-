```python
Encapsulation puts rules around your data so wrong values can never get in
```

```python
class Mobile:
    def __init__(self, brand, model, price, quantity):
        self.brand    = brand
        self.model    = model

        # YOUR CONDITION — price should be greater than 0
        if price > 0:
            self.price = price
        else:
            print("❌ Price cannot be 0 or negative! Setting default ₹1000")
            self.price = 1000

        # YOUR CONDITION — quantity should be greater than 0
        if quantity >= 0:
            self.quantity = quantity
        else:
            print("❌ Quantity cannot be negative! Setting default 0")
            self.quantity = 0
```

```python
m1 = Mobile("Samsung", "Galaxy S24", 80000, 10)
m2 = Mobile("Apple", "iPhone 15", -5000, -999)

print(m1.price)      # 80000   ✅ normal
print(m2.price)      # 1000    ✅ protected, default applied
print(m2.quantity)   # 0       ✅ protected, default applied
```

```python
""" the solution is 2 things together

1) make the attribute private - hide it from outside
2) create a controlled door - only way to change it is through your"""
```

```python
class Mobile1:
    def __init__(self, brand, model, price, quantity):
        self.brand = brand
        self.model = model
        self.__price = 0   # __ make it private
        self.__price = 0   # nobody can touch this directly

        self.set_price(price) # go through the controlled door
        self.set_quantity(quantity) # go through the controlled door

    def set_price(self, price):
        if price > 0:
            self.__price = price
        else:
            print('invalid price')
    def set_quantity(self, quantity):
        if quantity >= 0:
            self.__quantity = quantity
        else:
            print('invalid')
    def get_price(self):
        return self.__price 
    def get_quantity(self):
        return self.__quantity 
    def display(self):
        print(f"Brand: {self.brand}, Model: {self.model}, Price: ₹{self.__price}, Stock: {self.__quantity}")
    def avaiability(self):
        if self.__quantity > 0:
            print(f"{self.model} is available. {self.__quantity} units in stock.")
        else:
            print(f"{self.model} is OUT of stock.")

    def discount(self, percent):
        discount_amount = self.__price * percent / 100
        new_price       = self.__price - discount_amount
        print(f"After {percent}% discount, new price is ₹{new_price}")
```

```python
# Normal object
m1 = Mobile1("Samsung", "Galaxy S24", 80000, 10)
m1.display()
# Brand: Samsung, Model: Galaxy S24, Price: ₹80000, Stock: 10

# Try wrong values at creation
m2 = Mobile1("Apple", "iPhone 15", -5000, -999)
# ❌ Invalid price! Must be greater than 0
# ❌ Invalid quantity! Cannot be negative

# Try directly hacking after creation
m1.__price = -5000        # looks like it works...
print(m1.get_price())     # 80000 ✅ original is SAFE
                          # Python actually created a new separate variable
                          # the real __price inside is untouched

# Correct way to update price
m1.set_price(75000)       # ✅ goes through condition
m1.set_price(-100)        # ❌ Invalid price! condition blocks it
print(m1.get_price())     # 75000 ✅
```

```python
Scenario — Hospital Patient System
A hospital stores patient data. Every patient has —

a name
an age
a blood pressure reading (example: 120)
a doctor assigned to them

Rules to enforce —

age must be between 0 and 120
blood pressure must be between 60 and 200
name and doctor cannot be empty string

Methods the object should have —

display full patient details
check if blood pressure is normal (normal = between 90 and 140)
update blood pressure with validation
```

```python
1. Which attributes will be private?
2. Which attributes will be public?
3. How many setters do you need?
4. What condition will you put in blood pressure setter?
```

```python
class Hospital:
    def __init__(self, name, age, blood_pressure, doctor_name):
        self.__name = ""            # private
        self.__age = 0              # private  
        self.__blood_pressure = 0   # private
        self.__doctor_name = ""       # private 
        self.set_name(name)         # go through setter
        self.set_age(age)
        self.set_blood_pressure(blood_pressure)
        self.set_doctor_name(doctor_name)

    # GETTER - read only door

    def get_name(self):
        return self.__name
    def get_age(self):
        return self.__age 
    def get_blood_pressure(self):
        return self.__blood_pressure
    def get_doctor_name(self):
        return self.__doctor_name 
    
    # setters

    def set_name(self , name):
        if name != "":
            self.__name = name
        else:
            print('name cannot be empty')
    def set_doctor_name(self, doctor_name):
        if doctor_name != "":
            self.__doctor_name = doctor_name 
    def set_age(self , age):
        if age > 0 and age <= 120:
            self.__age = age
        else:
            print(f'invalid age {age}')
    def set_blood_pressure(self, blood_pressure):
        if blood_pressure >= 60 and blood_pressure <= 200:
            self.__blood_pressure = blood_pressure
        else:
            print(f'invalid BP {blood_pressure}')

        # Method
    def display(self):
        print(f'patient name - {self.__name}, age {self.__age} , BP {self.__blood_pressure}, doctor name {self.__doctor_name}')
    def is_dp_normal(self):
        if 90 <= self.__blood_pressure <= 140:
            print(f'{self.__name} is BP is normal')
        else:
            print(f'{self.__name} is BP is abnormal')      
```

```python
class Person:
    def __init__(self, name, age, phone_no):
        self.name = name
        self.age = age
        self.phone_no = phone_no
    def display(self):
        print(f'name - {self.name}, age = {self.age},phone no - {self.phone_no}')

class Teacher(Person):
    def __init__(self,name,age,phone_no,subject,salary):
        super().__init__(name,age,phone_no)
        self.subject = subject 
        self.salary = salary
    def teach(self):
        print(f'{self.name} teach this {self.subject}')
    def display(self):
        print(f'name - {self.name}, age = {self.age},phone no - {self.phone_no},subject - {self.subject}, salary - {self.salary}')
class Student(Person):
    def __init__(self, name, age, phone_no,grade,fees):
        super().__init__(name, age, phone_no)
        self.grade = grade
        self.fees = fees
    def result(self,grade):
        if grade >= 85 and grade <= 100:
            print(f'pass A {grade}')
        elif grade >= 70 and grade <=84:
            print(f'pass B {grade}')
        elif grade >= 35 and grade <=69:
            print(f'pass C {grade}')
        else:
            print(f'fail {grade}')
    def display(self):
        print(f'name - {self.name}, age = {self.age},phone no - {self.phone_no}, grade - {self.grade}, fees - {self.fees}')
```

```python
class Payment:
    def __init__(self,customer_name,amount):
        self.customer_name = customer_name
        self.amount = amount 
    def process(self):
        print('generic version')
class Creditcard(Payment):
    def __init__(self,customer_name,amount,card_no):
        super().__init__(customer_name,amount)
        self.card_no = card_no
    def process(self):
        print(f'Processing credit card payment of {self.amount}, for {self.customer_name}')
class UPI(Payment):
    def __init__(self,customer_name,amount,upi_id):
        super().__init__(customer_name,amount)
        self.upi_id = upi_id 
    def process(self):
        print(f'Processing UPI payment of {self.amount} to upi_id-{self.upi_id} for {self.customer_name}')
class NetBanking(Payment):
    def __init__(self,customer_name,amount,bank_name):
        super().__init__(customer_name,amount)
        self.bank_name = bank_name 
    def process(self):
        print(f'Processing Net banking payment of {self.amount} via {self.bank_name}')

payments = [Creditcard('ajit',20000,1324), UPI('amit',25000,2345), NetBanking('aman',30000,'Union bank')]
```

```python
for i in payments:
    i.process()
```
