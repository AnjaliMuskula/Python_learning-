```python
import random 

def get_score(attemts):
    if attemts >= 1 and attemts <=3:
        print(f'{attemts} Excellent')
    elif attemts >= 4 and attemts <= 6:
        print(f'{attemts} Good')
    elif attemts >= 7 and attemts <= 10:
        print(f'{attemts} Average')
    else:
        print(f'{attemts} keep practice')
    pass 
def play_game():
    num = random.randint(1,100)
    attemts = 0 
 

    while True:
        guess = int(input('enter your number:'))
        attemts += 1 

        if guess < num:
            print("to low")
        elif guess > num:
            print('to high')
        else:
            print(f' correct attemts {attemts}')
            get_score(attemts)
            break 
while True:
    play_game()
    again = input('Paly again (yes /no):')
    if again.lower() != 'yes':
        break
```

```python
num = random.randint(1, 100) is inside play_game()
because every time a NEW game starts —
a NEW random number must be generated.

If it was outside the function —
num is generated ONCE when program starts
every game would have the SAME number
play again would never feel like a new game



Key concept — variable scope. Inside function = fresh every call. ✅
```

```python
attempts = 0 is outside while loop because —
it needs to PERSIST across iterations
accumulate across every guess

If inside —
attempts = 0    ← resets every loop
attempts += 1   ← always stays 1
counter never grows — always shows 1 attempt

Outside —
attempt 1 → attempts = 1
attempt 2 → attempts = 2
attempt 3 → attempts = 3
grows correctly ✅


Key concept — variables outside loop persist, inside loop reset. ✅
```

```python
again.lower() 

User types "YES"  → .lower() → "yes" → matches "yes" ✅
User types "Yes"  → .lower() → "yes" → matches "yes" ✅
User types "YES"  → .lower() → "yes" → matches "yes" ✅

Without .lower() —
"YES" != "yes"  → loop breaks even though user said yes ❌

Key concept — always normalize string input before comparing. ✅ 


```

```python
if guess < num:
    print("too low")     # wrong guess → loop continues
elif guess > num:
    print("too high")    # wrong guess → loop continues
else:
    print("correct!")    # right guess → break stops loop
    break

No break after too low/high  →  loop continues  ✅
break only in else           →  stops when correct ✅

If break after every print —
game ends after first guess
whether correct or not ❌
```
