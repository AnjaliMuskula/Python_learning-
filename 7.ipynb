```python
name = input('enter your name:')
age = int(input("enter your age:"))
salary = int(input('enter your salary:'))
credit_score = int(input("enter your credit score:"))

age_ok = age >= 21 and age <=58 
salary_ok = salary >= 25000 
credit_ok = credit_score >= 700 

if age_ok and salary_ok and credit_ok:
    print(f'approved {salary * 24} loan')
elif not age_ok and salary_ok and credit_ok:
    print(f'{age} age requirement not met')
elif age_ok and not salary_ok and credit_ok:
    print(f'{salary} salary is low')
elif age_ok and salary_ok and not credit_ok:
    print(f'{credit_score} does not meet requirements')
else:
    print('does not meet any requirement')

print(f"test 1 name :{name}, age: {age}, salary: {salary}, credit {credit_score}")
# print(f"test 2 name :{name}, age: {age}, salary: {salary}, credit {credit_score}")
# print(f"test 3 name :{name}, age: {age}, salary: {salary}, credit {credit_score}")
# print(f"test 4 name :{name}, age: {age}, salary: {salary}, credit {credit_score}")
# print(f"test 5 name :{name}, age: {age}, salary: {salary}, credit {credit_score}")

```
