# Decorators

```python
# inshort Decorators --> Extra behavior

# Base --> your original function

# extra Layer --> behavior added on top 

# final result --> enhanced function --original still intact underseath 
```

```python
def decorator(func):    # accepts a function
    def wrapper():      # defines inner funcyion  
        print('before') # extra behavior before 
        func()          # calls original function
        print('after')  # extra behavior after 
    return wrapper      # returns wrapper
```

```python
decorator(func)
    │
    ├── receives original function as 'func'
    │
    ├── creates wrapper() — a new enhanced function
    │       │
    │       ├── print("Before...")
    │       ├── func()  ← original runs here
    │       └── print("After...")
    │
    └── returns wrapper — the enhanced version
```

```python
Write a decorator called log_it that —

prints "⏳ Function started..." before
runs the original function
prints "✅ Function finished." after

Apply it to —

process_payment(amount) → prints "Payment of ₹amount processed"
send_email(email) → prints "Email sent to email"
```

```python
def log_it(func):
    def wrapper(*args, **kwargs):
        print('function started......')
        func(*args,**kwargs)
        print('function finished')
    return wrapper

@log_it
def process_payment(amount):
    print(f'Payment of {amount} processed')

@log_it
def send_email(email):
    print(f'Email sent to {email}')

    
```

```python
process_payment(50000)
send_email('anjali.com')
```

```python
### recap --> decorators are used for add extra functions to a main function
```

# generators

# tenacious  - not give up

Normal Function      Generator
─────────────────────────────────────────────────────
keyword             return               yield
runs                all at once          one step at a time
memory              loads everything     one value at a time
remembers position  ❌ No               ✅ Yes
use case            small data           large/infinite data
real example        return a list        stream ChatGPT response

```python
# marks = [12,34,56,78,90]

def student_grade(markss):
    for mark in markss:
        if mark >= 35:
            yield mark
marks = [12,34,56,78,90]
for grade in student_grade(marks):
    print(grade)
```

```python
marks = [56, 78, 90]

# Step 1 — create iterator from list
it = iter(marks)          # calls __iter__() underneath

# Step 2 — get values one by one
print(next(it))           # 56  — calls __next__()
print(next(it))           # 78  — calls __next__()
print(next(it))           # 90  — calls __next__()
# print(next(it))           # ❌ StopIteration — nothing left
```

```python
# All of these are iterables — they have __iter__ underneath
for x in [1, 2, 3]:          # list
for x in (1, 2, 3):          # tuple
for x in {1, 2, 3}:          # set
for x in "Anjali":           # string — each character
for x in {"a": 1, "b": 2}:  # dictionary — each key
```

```python
Iterator        =  object that gives one value at a time
                   remembers position
                   knows when to stop

__iter__()      =  returns iterator object
__next__()      =  gives next value, raises StopIteration when done

iter()          =  creates iterator from any iterable
next()          =  gets next value manually

for loop        =  calls iter() and next() automatically underneath

Generator       =  easiest way to create an iterator using yield
```

```python
Iterable   →  anything that CAN be looped through
               list, string, dict, tuple

Iterator   →  the thing that DOES the one-by-one giving
               created from an iterable using iter()

Generator  →  special iterator using yield
```

# File Handling

```python
write --> create file , put data in
read --> open file , get data out 
append --> add more data without deleting existing 
```

```python
# structure 

with open('file_name.txt',"mode") as file:
    # do something with files 
```

```python
"w" -->write - create files , overwrites if exists 
"r" --> read - reads files , error if doesn't exist 
"a" --> append - adds to files , creates if doesn't exist 
"x" --> create files, error if alread exists
```

```python
with open('first_file','w') as file:
    file.write('anjali,22')
    file.write('Rahul,24')
```

```python
with open('first_file','r') as file:
    content = file.read()
    print(content)
```

```python
with open('first_file', 'a') as file:
    file.write('amit,25')
```

```python
with open('first_file','r') as file:
    a = file.read()
    print(a)
```

```python
#diference between write and append
write --> overwrite everything old data gone new data process_payment
append --> adds to existing data
```
