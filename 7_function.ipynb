```python
def greet(name):  #def = define, greet = name, name = parameter
"""say hello"       #docstring - explains what function does 
result = f'hello' {name}!'
return result             # sends value back to whoever called it

message = greet('anjali')   #calling the function
print(message)              # hello arjun
```

```python
def = tells python 'a function is coming'
greet = the name you call it by
(name) = input the function expects
return = output the function gives back 
```
