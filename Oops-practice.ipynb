# Encapsulation 

```python
1) write class bankAccount 

2) attributes --> account holder name , balance , account type(savings/ current)

3) Rules

balance cannot be negative
name cannot be empty
account type must be either "savings" or "current" 

4) Methods 

deposit(amount) — add money, amount must be positive
withdraw(amount) — remove money, can't withdraw more than balance
display() — show all details
```

```python
class BankAccount():
    def __init__(self, name, balance, account_type):
        self.__name = "" 
        self.__balance = 0 
        self.__account_type = ""  
        self.set_name(name)   #setter go through setter
        self.set_balance(balance)   #  
        self.set_account_type(account_type)

    #### getter
    def get_name(self):
        return self.__name 
    def get_balance(self):
        return self.__balance 
    def get_account_type(self):
        return self.__account_type 

    ### setter
    def set_name(self,name):
        if name != "":
            self.__name = name 
        else:
            print('enter name first')
    def set_balance(self, balance):
        if balance > 0:
            self.__balance = balance 
        else:
            print('balance cant be negative ')
    def set_account_type(self, account_type):
        if account_type == 'saving':
            self.__account_type = account_type 
        elif account_type == 'current':
            self.__account_type = account_type 
        else:
            print(f'invalid {account_type}account type ')

    ### Methods 
    
    def deposit(self,amount):
        if amount > 0:
            self.__balance = self.__balance + amount
            print(f'amount {amount} deposited ! New balance {self.__balance}')
        else:
            print(f'invalid amount {amount}')
    def withdraw(self,amount):
        if amount < self.__balance:
             self.__balance = self.__balance - amount
             print(f'balance amount {amount}')
        else:
            print(f'invalid amount {amount}')
    def display(self):
        print(f'name - {self.__name}, balance  - {self.__balance}, account_type - {self.__account_type}')

    
    



```

```python
a3 = BankAccount('Aurn', 28000,'saving')
a3.display()
a3.deposit(1000)
a3.withdraw(2000)
a3.display()

```

```python
class Employee():
    def __init__(self,name,department,salary, emp_id):
        self.__name = ""
        self.__department = ""
        self.__salary = 0
        self.__emp_id = 0 
        self.set_name(name)
        self.set_department(department)
        self.set_salary(salary)
        self.set_emp_id(emp_id)

    # getter 
    def get_name(self):
        return self.__name 
    def get_dep(self):
        return self.__department 
    def get_salary(self):
        return self.__salary 
    def get_emp_id(self):
        return self.__emp_id 
    
    # setter condition 

    def set_name(self,name):
        if name != "":
            self.__name = name 
        else:
            print('name cannot be empty please enter name ')
    def set_department(self,department):
            if department in ['IT', 'HR', 'Finance']:
                self.__department = department 
            else:
                print('invalid')
        
            
    def set_salary(self , salary):
        if salary >= 10000 and salary <= 500000:
            self.__salary = salary 
        else:
            print('enter salary between 10000 to 500000')
    def set_emp_id(self, emp_id):
        if emp_id >= 1000 and emp_id <= 9999:
            self.__emp_id = emp_id
        else:
            print('Invalid emp ID')
    ## Methods 

    def give_raise(self,raise_amount):
        new_salary = self.__salary + raise_amount
        if new_salary > 500000: 
            print('raise exceed limit')
        else:
            self.__salary = new_salary 
            print(f'new_salary : {self.__salary}')

    def display(self):
        print(f'name-{self.__name}, dep - {self.__department}, salary - {self.__salary}, emp_id - {self.__emp_id}')
        
```

```python
e1 = Employee('ABC','IT',30000,1202)
```

```python
e1.give_raise(10000)
```
