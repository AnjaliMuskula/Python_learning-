| Exception           | Meaning                   |
| ------------------- | ------------------------- |
| `ValueError`        | Wrong value type          |
| `TypeError`         | Wrong data type operation |
| `ZeroDivisionError` | Divide by zero            |
| `IndexError`        | List index out of range   |
| `KeyError`          | Dictionary key not found  |
| `FileNotFoundError` | File doesn't exist        |
| `NameError`         | Variable not defined      |


ValueError      →  wrong type  int("abc")
ZeroDivisionError → divide by 0  10/0
FileNotFoundError → file missing  open("abc.txt")
IndexError      →  list index out of range  list[99]
KeyError        →  dict key missing  dict["wrong_key"]
TypeError       →  wrong operation  "text" + 5
Exception       →  catches ALL errors  ← use as last resort

```python
Exception Handling is a mechanism used to handle runtime errors in a program 
using try, except, else, and finally blocks, allowing the program to continue 
execution without crashing.
```

### syntax

```python
try:
    # code that might fail 
    # if error occurs --> jumps to except immediately 
except ErrorType:
    # run only if error accurred 
    # handle it gracefully 
else:
    # runs only if NO error occurred 
    # optional 
finally:
    # always runs -- error or not 
    # used for cleanup 
    # optional 
```

```python
# Value Error 

try:
    num = int('acb')
except ValueError:
    print('Not a valid number')


 # user enter invalid input    
```

```python
# ZeroDivisionError 

try:
    result = 10/0
except ZeroDivisionError:
    print('cannot divide by zero')
```

```python
#fileNot founderror 

try:
    with open('missing.txt','r') as f:
        print(f.read())
except FileNotFoundError:
    print('file is missing')
```

```python
# Multiple error at once 

try:
    num = int(input('enter number'))
    result = 10 / num
except ValueError:
    print('invalid number')
except ZeroDivisionError:
    print('cannot divide by zero')
print(result)
```

### Finally -- Always Runs 

```python
try:
    file = open('data.txt','r')
    content = file.read()
except FileNotFoundError:
    print('file not found')
finally:
    print('program finished')
```

If file exists   →  reads + "Program finished" 
If file missing  →  error message + "Program finished"

# Expense Tracker 

```python
import json 
import os 

def load_expenses():
    if os.path.exists('expenses.json'):
        with open('expenses.json','r') as file:
            return json.load(file)
    return []  # empty list if file doesn't exist 

# step 2 show menu 

def show_menu():
    print('\n-------Expenses Tracker-------')
    print('1.add Expenses')
    print('2.View all')
    print('3. total by category')
    print('4. Exit')
    return input('choice:')

def save_expenses(expenses):
    with open("expenses.json", "w") as file:
        json.dump(expenses, file, indent=4)


def log_error(error):
    with open("errors.log", "a") as file:
        file.write(f"ERROR: {error}\n")

# main

expenses = load_expenses()

while True:
    choice = show_menu()

    if choice == '1':
        try:
            name = input('expenses name:')
            if name == "":
                raise ValueError('Name cannot be empty')
            
            amount = float(input('amount :'))
            if amount <= 0:
                raise ValueError('amount must be positive')
            
            
            print('categories: Food, Transport, Shopping, Bills, Other')
            category = input('category:')
            if category not in ['Food','Transport','Shopping','Bills','Other']:
                raise ValueError(f'Invalid category: {category}')
            
            expense = {"name": name
                       , "amount": amount,
                       "category": category
            }
            expenses.append(expense)
            save_expenses(expenses)
            print(f"✅ {name} ₹{amount} added!")

        except ValueError as e:
            print(f"❌ {e}")
            log_error(str(e))
    elif choice == '2':
        if len(expenses) == 0:
            print("No expenses yet!")
        else:
            print("\n─── All Expenses ───")
            for i, exp in enumerate(expenses, 1):
                print(f"{i}. {exp['name']} ₹{exp['amount']} - {exp['category']}")

    elif choice == '3':
        if len(expenses) == 0:
            print("No expenses yet!")
        else:
            print("\n─── Total by Category ───")
            categories = ["Food", "Transport", "Shopping", "Bills", "Other"]
            for cat in categories:
                total = sum(exp['amount'] for exp in expenses if exp['category'] == cat)
                if total > 0:
                    print(f"{cat}: ₹{total}")
    elif choice == '4':
        print("Goodbye! 👋")
        break

    else:
        print("❌ Invalid choice!")


            
```
