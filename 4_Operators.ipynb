# Aritmetic Operation

| Operator | Meaning             |
| -------- | ------------------- |
| `+`      | Addition            |
| `-`      | Subtraction         |
| `*`      | Multiplication      |
| `/`      | Division            |
| `//`     | Floor Division      |
| `%`      | Modulus (remainder) |
| `**`     | Exponent (power)    |


```python
### arithmetic operation --> it is simply used for calculation 

num1 = int(input('enter your number1'))
num2 = int(input('enter your number2'))


add = num1 + num2
print(f'addition of {num1} and {num2} is {add}')

sub = num1 -  num2
print(f'subtract of {num1} and {num2} is {sub}')

multi = num1 * num2
print(f'multiplication of {num1} and {num2} is {multi}')

division = num1 / num2  # value = int + float
print(f'division of {num1} and {num2} is {division}')

floor = num1 // num2   # value = int
print(f'floor division {num1} and {num2} is {floor}')
```

```python
#Q2) ask user for base number give power

user = int(input('enter a number'))

power = user**2
print(power)
```

```python
#Q3) ask user for total chocolate and friends

choco = int(input('enter how many chocolate'))
friends = int(input('enter your friends'))

each_choco = choco // friends
print(f'each chocolate will get {each_choco}')

remaining = choco % friends
print(f'remaining {remaining}')
```

Comparison Operators

| Operator | Meaning               |
| -------- | --------------------- |
| `==`     | Equal to              |
| `!=`     | Not equal             |
| `>`      | Greater than          |
| `<`      | Less than             |
| `>=`     | Greater than or equal |
| `<=`     | Less than or equal    |


```python
## comparison opeartors is simply used to compory the values

num = int(input('enter your age'))

if num>=18:
    print('eligiable')
else:
    print('not eligiable')
```

```python
# Q3) ask user for 3 numbers

num1 = int(input('enter num1'))
num2 = int(input('enter num2'))
num3 = int(input('enter num3'))

if num1 > num2  and num1 > num3:
    print(f'number {num1} is greater than {num2} and {num3}')

elif num2 > num1 and num2 > num3:
    print(f'number {num2} is greater than {num1} and {num3}')
elif num3 > num1 and num3> num2:
    print(f'number {num3} is greater than {num2} and {num1}')
else:
    print(f'all are same {num1}, {num2} and {num3}')
```

```python
#Q3) ask user for 2 numbers  check if both are equal

num1 = int(input('enter a number'))
num2 = int(input('enter a number'))

if num1==num2:
    print(f'{num1} ,{num2} both are same')
elif num1 != num2:
    print(f'{num1} , {num2} are not equal')

```

# Membership operators

```python
# membership operators is simply used to check does data exist or not
```

| Operator | Meaning     |
| -------- | ----------- |
| `in`     | Present     |
| `not in` | Not present |


```python
name = 'anjali'

user = input('enter a letter')

if user in name:
    print(f'{user} is present in {name}')
else:
    print('not present')
```

```python
l = ['python','c++','c']

user = input('enter your coding language')
if user not in l:
    print(f'{user} is not present')
else:
    print('its present')
```
