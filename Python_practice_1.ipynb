### Dict --> key value pair

```python
student = {'name':'Anjali',
           'age': 22,
           'marks':92,
           'city':'mumbai'}
# acces 
print(student['name'])
print(student.get('age'))

# modify
student['marks'] = 99

# delete 
del student['city']

# loop through 
for key, value in student.items():
    print(f'{key} : {value}')

```

### List Comprehension -->powerful one linear

```python
# normal way
marks = [12,34,67,89]
passed = []

for i in marks:
    if i >30:
        passed.append(i)
    else:
        print('fail')
print(passed)


#list comprehensive 

passed = [m for m in marks if m > 30]
print(passed)
```

```python
Given this list —
marks = [45, 92, 38, 78, 55, 29, 88, 61, 73, 40]

Using list comprehension —
1. Create a list of only passed students (marks >= 40)
2. Create a list of marks doubled for passed students only
3. Print highest, lowest, average of passed marks
```

```python
marks = [45, 92, 38, 78, 55, 29, 88, 61, 73, 40]

pas = [m for m in marks if m >= 40]
print(pas)

double = [m * 2 for m in pas]
print(double)

# rank = [m for m in pas if m <= 50 'lowest', elif m <= 70 'average', else 'highest']

print('highest',max(pas))
print('lowest',min(pas))
print('average',sum(pas))
```

### List comprehension is for filtering or transforming — not for complex if/elif logic

```python
Given this list with duplicates —
students = ["Anjali", "Rahul", "Priya", "Anjali", 
            "Rahul", "Amit", "Priya", "Neha"]

1. Remove all duplicates using set
2. Store each student with a default marks of 0 in a dict
3. Update Anjali's marks to 91, Rahul's to 78
4. Print all students and their marks
```

```python
students = ["Anjali", "Rahul", "Priya", "Anjali", 
            "Rahul", "Amit", "Priya", "Neha"]

unique = set(students)
print(unique)

student_marks = {}

for i in unique:
    student_marks[i] =0 
print(student_marks)

student_marks['anjali'] = 91
print(student_marks.get('anjali'))

for name , value in student_marks.items():
    print(f'{name} : {value}')
```

```python
You are building a system for a cricket team.

Given this list of players with duplicates —

players = ["Rohit", "Virat", "Rohit", "Dhoni",
           "Virat", "Bumrah", "Dhoni", "Jadeja"]

1. Remove duplicate players using set

2. Store each player in a dict with 
   default score of 0

3. Update these scores —
   Rohit  → 85
   Virat  → 120
   Dhoni  → 67

4. Using a loop — print only players 
   whose score is greater than 0
   format → "Virat scored 120 runs ✅"

5. Find the highest scorer and print —
   "Best player: Virat with 120 runs 🏆"
```

```python
players = ["Rohit", "Virat", "Rohit", "Dhoni",
           "Virat", "Bumrah", "Dhoni", "Jadeja"]

# # 1. Remove duplicate players using set
unique = set(players)
print(f' 1.{unique}')

# #2 Store each player in a dict with default score of 0
score = {}

for i in unique:
    score[i] = 0
print(f'{score}')

# # 3. Update these scores —Rohit  → 85 Virat  → 120 Dhoni  → 67
score['Rohit'] = 85
score['Virat'] = 120 
score['Dhoni'] = 97 

# # 4 Using a loop — print only players whose score is greater than 0 format → "Virat scored 120 runs ✅"

for i in score:
    if score[i] > 0:
        print(f'3 {i} scored {score[i]}')
    
# #Find the highest scorer and print —"Best player: Virat with 120 runs 🏆"


best1 = max(score,key=score.get)
print(f'best player: {best1} with {score[best1]} runs')

```

```python
| What are you looping over? | Example                      |
| -------------------------- | ---------------------------- |
| List                       | `for x in my_list`           |
| String                     | `for ch in text`             |
| Range of numbers           | `for i in range(10)`         |
| Tuple                      | `for x in my_tuple`          |
| Set                        | `for x in my_set`            |
| Dictionary keys            | `for k in d`                 |
| Dictionary values          | `for v in d.values()`        |
| Dictionary items           | `for k, v in d.items()`      |
| File lines                 | `for line in file`           |
| Multiple lists             | `for a, b in zip(...)`       |
| Index + value              | `for i, x in enumerate(...)` |
| Condition-based repetition | `while condition`            |
| Infinite repetition        | `while True`                 |
```
