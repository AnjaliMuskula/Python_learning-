```python
a = ['sum','add','div']

for i in a:
    print(i.upper())

 # go to a
 # pick up the first item --> 'sum'
 # run the block --> print('SUM')
 # Go back to a --> pick up next item --> i = 'add'
 # Run the block --> print('add')
 # repeat until no more item
 # exit the loop
```

```python
i = 1   #start

while i <= 5:  # condition
    print(i)   # print
    i += 1     # iterates each position

    # i+= 1
    # updation
    # without it = infinite loop
```

```python
#find sum of 1 to 100
for i in range(1,100):
    sum = sum + i
print(sum)  # print statement is outside the loop so it will be the final answer
# if print statement inside the loop so the output will be loops

# print iside ==> prints n times -- once for each value of i
# print outside ==> prints once --after the loop is completely done 
```

```python
sum = 0  ← empty bucket

loop starts:
  i=1  → throw 1  into bucket → bucket has 1
  i=2  → throw 2  into bucket → bucket has 3
  i=3  → throw 3  into bucket → bucket has 6
  ...
  i=100 → throw 100 into bucket → bucket has 5050

loop ends → look inside bucket → 5050
```

```python
for i in range(1, 6):
    sum += i
    print(sum)   # 👈 inside = "show me DURING the work"

print(sum)       # 👈 outside = "show me AFTER the work is done"
```
