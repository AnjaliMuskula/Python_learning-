# Project --> Student Report Card

```python
name = input("enter your name:")
age = int(input("enter your age:"))
marks = int(input("enter your marks:"))

# find percentage 

percentage = marks / 150 * 100

# find grade 
if marks >= 120:
    grade = 'A'
elif marks >= 80:
    grade = 'B'
elif marks >= 60:
   grade = 'C'
elif marks >= 45:
    grade = 'D'
else:
    grade = 'E'

# find pass or fail 
if marks >= 45:
    result = 'pass'
else:
    result = 'Fail'

# step 5 print 
print("----------------------")
print('student report card')
print(f'Name: {name}')
print(f'age: {age}')
print(f'Marks: {marks}')
print(f'Percentage: {percentage}%')
print(f'Grade: {grade}')
print(f'Result: {result}')

# save to file

with open('report_card.txt','w') as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
    file.write(f"Marks: {marks}\n")
    file.write(f'percentage: {percentage}\n')
    file.write(f'Grade: {grade}\n')
    file.write(f'Result:{result}')
# ... and so on)
```

```python
Write a program that —
takes a temperature in Celsius as input
converts to Fahrenheit and Kelvin
prints all 3 values formatted nicely

Formula —
Fahrenheit = (Celsius * 9/5) + 32
Kelvin     = Celsius + 273.15
```

```python
celsius = float(input('enter celsius:'))

fah = (celsius * 9/5) + 32
kel = celsius + 273.15 

print(f'celsius: {celsius}')
print(f'fehrenheit: {fah}')
print(f'kelvin: {kel}')
```
