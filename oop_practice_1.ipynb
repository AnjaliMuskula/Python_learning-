# 2 Encapsulation 

```python
### safe , private , and condition on data
```

```python
class Students:
    def __init__(self, name , age, marks):
        self.__name = ""
        self.__age = 0
        self.__marks = 0
        self.set_name(name) 
        self.set_age(age)
        self.set_marks(marks) 


    # getter 
    def get_name(self):
        return self.__name 
    def get_age(self):
        return self.__age 
    def get_marks(self):
        return self.__marks  
    
    # setter 
    def set_name(self, name):
        if name != "":
            self.__name = name 
        else:
            print('name cannot be empty')
    def set_age(self, age):
        if 5 <= age <= 30:
            self.__age = age
        else:
            print(f'invalid age: {age}')
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print(f'Invalid marks {marks}')
    def display(self):
        print(f'name : {self.__name}, age: {self.__age} , marks: {self.__marks}')

```

```python
s2 = Students('amit',12,23)
s2.display()
```

```python
s3 = Students("",23,67)
s3.display()
s3.get_age()
```

# 3 Inheritance 

### inherit something from parents to child 

```python
class Person:
    def __init__(self, name, age):
        self.name = name 
        self.age = age 

    def display(self):
        print(f'name: {self.name}, age: {self.age}')

class Student(Person):
    def __init__(self,name,age,marks):
        super().__init__(name, age)
        self.marks = marks 

    def study(self):
        print(f'name: {self.name}, age: {self.age}, marks: {self.marks}')

```

```python
p1 = Person('sunio',23)
p1.display()

stu = Student('ajit',2,9)
stu.display()
```

```python
# Person = parent - common things (name, age)
# student = child - inherits name , age + add marks

# super() = calls parents __init__ first 
```

# 4 Polymorphism

### you want to do same work but different action 

```python
class Person:
    def __init__(self, name):
        self.name = name 
    def work(self):
        print(f'{self.name} is working')
class Student(Person):
    def work(self):
        print(f'{self.name} is studing')
```

```python
people = [Person('anjali'),Student('ajit')]
```

```python
for i in people:
    i.work()
```

Person (Parent)
├── name, age
├── display()
└── work()  ← base version
      │
      ├── Student(Person)
      │     ├── +marks          ← encapsulation protects it
      │     ├── study()         ← own method
      │     └── work()          ← polymorphism override
      │
      ├── Teacher(Person)
      │     ├── +subject        ← encapsulation protects it
      │     ├── teach()         ← own method
      │     └── work()          ← polymorphism override
      │
      └── Principal(Person)
            ├── +office         ← own attribute
            └── work()          ← polymorphism override

```python
Class          =  skeleton — defines structure
Object         =  real thing built from skeleton
Encapsulation  =  lock data + add rules on door
Inheritance    =  child gets parent's qualities + adds own
Polymorphism   =  same action, different behavior per object
super()        =  child calling parent's __init__
__attribute    =  private — locked inside class
getter         =  read the private data
setter         =  write with conditions
```

# School System 

```python
class Person:
    def __init__(self,name,age):
        self.__name = ""
        self.__age = 0
        self.set_name(name)
        self.set_age(age)

    # getter 
    def get_name(self):
        return self.__name 
    def get_age(self):
        return self.__age 
    
    # setter 
    def set_name(self,name):
        if name != "":
            self.__name =name 
        else:
            print('name cnnot be empty')
    def set_age(self, age):
        if age > 0:
            self.__age = age 
        else:
            print('invalid age: {age}')
    def display(self):
        print(f'name {self.__name}, age {self.__age}')

    def work(self):
        print(f'{self.__name} is a person')

class Student(Person):
    def __init__(self, name, age, marks,grade):
        super().__init__(name,age)
        self.__marks = 0
        self.__grade = 0 
        self.set_marks(marks)
        # self.set_grade(grade)

    # getter
    def get_marks(self):
        return self.__marks
    def get_grade(self):
        return self.__grade 
    
    # setter 

    def set_marks(self,marks):
        if marks != 0:
            self.__marks = marks 
        else:
            print(f'invalid marks {marks}')
    def study(self):
        print(f'{self.get_name()} is studing')
    def work(self):
        print(f'{self.get_name()} is a student')

class Teacher(Person):
    def __init__(self,name, age, subject, salary):
        super().__init__(name,age)
        self.__subject = ""
        self.__salary = 0 
        self.set_sub(subject)
        self.set_salary(salary)

    # getter 
    def get_sub(self):
        return self.__subject
    def get_salary(self):
        return self.__salary 
    def set_sub(self, subject):
        if subject != "":
            self.__subject = subject

    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
    
    def teach(self):
        print(f'{self.get_name()} teaches {self.__subject}')

    def work(self):
        print(f'{self.get_name()} is a teacher')

class Principle(Person):
    def __init__(self,name,age,school_name):
        super().__init__(name,age)
        self.__school_name= ""
        self.set_school_name(school_name)

    def set_school_name(self, school_name):
        if school_name != "":
            self.__school_name = school_name
    def work(self):
        print(f'{self.get_name()} is a principle in {self.__school_name}')



 
```

```python
p1 = Person('pramila',67)
p1.display()
p1.work()

s1 = Student('ajit',13,19,'A')
s1.study()
s1.work()

t1 = Teacher('Akshara', 30,'English',30000)
t1.teach()
t1.work()

p2 = Principle('Suraj',65,'Gurukul')
p2.work()
```

```python
people = [
    Student("Anjali", 23, 91, "A"),
    Teacher("Mr. Sharma", 45, "Maths", 50000),
    Principle("Mrs. Mehta", 50, "DPS School")
]

for person in people:
    person.work()
```
