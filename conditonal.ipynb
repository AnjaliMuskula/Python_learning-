```python
## easy level
# ask user for a number
user = int(input('enter a number'))
if user % 2 == 0:
    print(f'even number {user}')
else:
    print(f'odd number {user}')
```

```python
# positive / negative / zero
user = int(input('enter a number'))
if user > 0:
    print('positive')
elif user < 0:
    print('negative')
else:
    print('zero')
```

```python
# voting eligibility
user = int(input('enter your age:'))
if user >= 18:
    print('eligible')
else:
    print('not eligible')
```

```python
# largest of two number 

num1 = int(input('enter a number'))
num2 = int(input('enter a number'))

if num1==num2:
    print('both are same')
elif num1 > num2:
    print(f'{num1} is greater than {num2}')
else:
    print(f'{num2} is greater than {num1}')
```

```python
# grade calculator
user = int(input('enter your number'))

if user >= 90:
    print('A')
elif user >=70 and user <=89:
    print('B')
elif user >=50 and user <= 69:
    print('c')
else:
    print('fail')
```

```python
# leap year checker
year = int(input('enter a year'))

if year % 4 == 0:
    print('leap year')
else:
    print('not leap year')
```

```python
# number range checker
range = int(input('enter in range'))
if range >= 10 and range <= 50:
    print('in range')
else:
    print('out in range')
```

```python
# password checker
password = input('enter a password')
if password =='python123':
    print('access granted')
else:
    print('not access')
```

```python
# simple atm
balance = int(input('enter your balanace'))
withdraw = int(input('enter with draw'))
if withdraw > balance:
    print('insufficient balance')
else:
    remainig_balanace = balance - withdraw
    print(f'{remainig_balanace}')
```

```python
# mini login system
user1 = 'anjali'
password1 = 123

user = input('enter your username:')
password = int(input('enter your password:'))

if user==user1 and password == password1:
    print('login successfull')
elif user != user1 and password == password1:
    print('incorrect user name')
else:
    print('user not found')
```

```python
# largest of 3 numbers
num1 = int(input('enter a number'))
num2 = int(input('enter a number'))
num3 = int(input('enter a number'))

if num1 > num2 and num1 > num3:
    print(f'{num1}')
elif num2 > num1 and num2 > num3:
    print(f'{num2}')
elif num3 > num1 and num3 > num2:
    print(f'{num3}')
else:
    print('all same')
```
